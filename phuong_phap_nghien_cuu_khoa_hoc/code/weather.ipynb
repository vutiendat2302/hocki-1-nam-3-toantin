```python
from datetime import datetime
import pandas as pd
from meteostat import Point, Daily

# 1. Định nghĩa Tọa độ và Khoảng thời gian
# Tọa độ Hà Nội (ví dụ: Trung tâm)
LATITUDE_HANOI = 21.0285
LONGITUDE_HANOI = 105.8542

# Khoảng thời gian lấy dữ liệu (Lấy hết năm 2024)
start_date = datetime(2024, 1, 1)
end_date = datetime(2025, 1, 1) # Meteostat sẽ lấy dữ liệu cho đến ngày 31/12/2024

# 2. Định nghĩa Điểm (Point)
hanoi = Point(LATITUDE_HANOI, LONGITUDE_HANOI)

# 3. Lấy dữ liệu Thời tiết Hàng ngày (Daily)
try:
    data = Daily(hanoi, start_date, end_date)
    df_weather = data.fetch()

    # 4. Lưu dữ liệu vào file CSV
    OUTPUT_FILE = 'hanoi_weather_2024.csv'
    df_weather.to_csv(OUTPUT_FILE)

    print(f"✅ Đã tải và lưu dữ liệu thời tiết Hà Nội vào file '{OUTPUT_FILE}' thành công.")
    print("\nThông tin 5 dòng đầu của dữ liệu:")
    print(df_weather.head())
    
    # In cảnh báo nếu có lỗi tải (ví dụ: lỗi VGGL0, nhưng vẫn có dữ liệu)
    # Dữ liệu bạn đã tải trước đó vẫn có 5 dòng đầu, chứng tỏ việc lấy data thành công.
    #
    
except Exception as e:
    print(f"❌ Lỗi xảy ra trong quá trình lấy hoặc lưu dữ liệu: {e}")
```
